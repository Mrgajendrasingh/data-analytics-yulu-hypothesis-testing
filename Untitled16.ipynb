**Step 15 — Formulate our first hypothesis test**

```python
from scipy.stats import ttest_ind

non_working = yulu.loc[yulu['workingday'] == 0, 'count']
working = yulu.loc[yulu['workingday'] == 1, 'count']

t_stat, p_value = ttest_ind(
    non_working,
    working,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Hypothesis test: Welch's independent two-sample t-test.

H₀: The population mean hourly rental counts are equal on working and non-working days.

H₁: The population mean hourly rental counts differ between working and non-working days.

Results: \(t=-1.326\), \(p=0.2164\), and \(\alpha=0.05\).

Decision: Fail to reject H₀ because the p-value exceeds the significance level.

Conclusion: The test does not provide sufficient evidence of a difference in population mean hourly rental counts between working and non-working days.

**Step 16 — Next: ANOVA across seasons**

```python
yulu.groupby('season', observed=True)['count'].agg(
    ['count', 'mean', 'median', 'std']
).round(2)
```

```python
from scipy.stats import f_oneway

groups = [
    yulu.loc[yulu['season'] == s, 'count']
    for s in [1, 2, 3, 4]
]

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

**conclusion: **One-way ANOVA found a statistically significant difference in mean hourly rental counts across the four seasons (\(F \approx 236.95,\ p < 0.001\)).

# Phase 3: Hypothesis Testing

**1. Independent Two-Sample t-Test**

Business question: Is mean hourly bike rental demand different between working days and non-working days?
Hypotheses
- \(H_0\): Mean rental counts are equal between the two groups.
- \(H_1\): Mean rental counts are different between the two groups.
Significance level: \(\alpha = 0.05\)

```python
from scipy.stats import ttest_ind

non_working = yulu.loc[yulu['workingday'] == 0, 'count']
working = yulu.loc[yulu['workingday'] == 1, 'count']

t_stat, p_value = ttest_ind(
    non_working, working, equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Results:

Metric Result:
Non-working day mean=188.51

Working day mean=193.01

t-statistic	-1.236

p-value	0.2164

Significance level	0.05

Decision and interpretation

Since \(p = 0.2164 > 0.05\), fail to reject \(H_0\).

Business interpretation: The observed mean rental count is slightly higher on working days, but the test does not provide sufficient evidence of a statistically significant difference in population means.

Assumptions
- Two independent groups.
- Numerical outcome (count).
- Large samples support approximate inference despite non-normality.
- Welch's t-test does not require equal variances.
- Caution: Hourly rental observations may be correlated over time, which can violate the independence assumption.

**Status**
Test 1 complete. Next is Test 2: One-way ANOVA — rental counts across weather categories.

**2: One-way ANOVA — rental counts across weather categories.**

1. Hypotheses
- H₀: Mean rental counts are equal across weather categories.
- H₁: At least one weather category has a different mean rental count.
- Significance level: \(\alpha = 0.05\)
2. Assumptions
1. Independent observations.
2. Numerical outcome: count.
3. Approximately normal residuals within groups, or sufficiently robust inference.
4. Equal population variances for standard one-way ANOVA.
Important: Weather category 4 has only one observation, so we'll perform the primary ANOVA on categories 1–3 and document this limitation.

```python
from scipy.stats import f_oneway, levene

weather_groups = [
    yulu.loc[yulu['weather'] == w, 'count']
    for w in [1, 2, 3]
]

# Check variance assumption
lev_stat, lev_p = levene(*weather_groups)

# Perform ANOVA
f_stat, p_value = f_oneway(*weather_groups)

print("Levene p-value:", lev_p)
print("F-statistic:", f_stat)
print("ANOVA p-value:", p_value)
```

**Test 2 — Results**

Significance level	0.05

Decision: Ordinary ANOVA rejects \(H_0\), since \(p < 0.05\).
Important assumption check: Levene's p-value is below 0.05, indicating unequal group variances.
So let's verify the result using Welch's ANOVA, which does not assume equal variances.

**3. One-way ANOVA: rental counts across seasons.**

```python
from statsmodels.stats.oneway import anova_oneway

weather_data = yulu[yulu['weather'].isin([1, 2, 3])]

result = anova_oneway(
    data=weather_data['count'],
    groups=weather_data['weather'],
    use_var='unequal'
)

print(result)
```

Business interpretation:

Average rental demand differs significantly across weather categories 1–3. Category 1 has the highest observed mean, while category 3 has the lowest. This establishes an association, not causation.

Assumptions:

Levene’s test indicated unequal variances, which is why Welch’s ANOVA was appropriate. Independence remains an important assumption. Weather category 4 was excluded because it contains only one observation.

**Test 4 — Chi-square test: Weather vs. Season**

Question: Are weather category and season independent?
- \(H_0\): Weather and season are independent.
- \(H_1\): Weather and season are associated.
- Significance level: \(\alpha=0.05\)

```python
from scipy.stats import chi2_contingency

# Exclude weather category 4 (only one observation)
data = yulu[yulu['weather'].isin([1, 2, 3])]

contingency_table = pd.crosstab(
    data['weather'],
    data['season']
)

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Contingency table:")
print(contingency_table)

print("\nChi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("P-value:", p_value)
print("Minimum expected frequency:", expected.min())
```

Hypotheses
- \(H_0\): Weather and season are independent.
- \(H_1\): Weather and season are associated.

Assumptions:

Categorical variables, independent observations, and sufficiently large expected cell frequencies. The expected-frequency condition is satisfied.
Decision: Since \(p < 0.05\), reject \(H_0\).
Business interpretation: There is statistically significant evidence of an association between weather category and season. This does not establish causation.

**Verify the working-day t-test**

```python
from scipy.stats import ttest_ind

non_working = yulu.loc[yulu['workingday'] == 0, 'count']
working = yulu.loc[yulu['workingday'] == 1, 'count']

print("Non-working days:")
print("Count:", len(non_working))
print("Mean:", non_working.mean())
print("Std:", non_working.std())

print("\nWorking days:")
print("Count:", len(working))
print("Mean:", working.mean())
print("Std:", working.std())

t_stat, p_value = ttest_ind(
    non_working, working, equal_var=False
)

print("\nT-statistic:", t_stat)
print("P-value:", p_value)
```

**Business interpretation:**

We don't have sufficient statistical evidence of a difference in mean hourly rentals between working and non-working days. The observed means are similar, with working days slightly higher.

# Phase 4: Business Insights and Recommendations

**Seasonal demand:** One-way ANOVA found a statistically significant difference in mean rental counts across seasons (F = 236.95, p < 0.001). Season 3 had the highest observed mean (234.42), while season 1 had the lowest (116.34).

**Weather and demand:** Welch's ANOVA found a statistically significant difference in mean rental counts across weather categories 1–3 (F = 140.90, p < 0.001). Weather category 1 had the highest observed mean (205.24), while category 3 had the lowest (118.85).

**Weather and season:** The chi-square test found a statistically significant association between weather category and season (χ² = 46.10, df = 6, p < 0.001).

**Outliers:** The IQR method flagged 300 hourly rental counts above the upper boundary of 647. These observations may represent genuine periods of high demand and should not automatically be removed.

# **2. Business Recommendations**

1. **Plan for seasonal demand:** Use seasonal rental patterns to support bike availability, maintenance scheduling, and operational planning.

2. **Monitor weather conditions:** Track rental demand by weather category to improve short-term demand planning and bike distribution.

3. **Investigate peak-demand periods:** Examine high-rental hours to identify recurring demand patterns and potential opportunities to improve bike availability.

4. **Avoid relying on working-day status alone:** The t-test did not find a statistically significant difference in mean rental counts. Combine working-day information with time, season, and weather when planning operations.

5. **Validate recommendations with further analysis:** Evaluate hourly and daily trends and test whether findings remain consistent across different periods.


# **3. Limitations**

- The analysis identifies statistical associations and differences but does not establish causation.
- Hourly observations may be correlated over time, which can affect tests that assume independent observations.
- The standard ANOVA variance assumption was violated for weather groups, so Welch's ANOVA was used.
- Weather category 4 contained only one observation and was excluded from the weather ANOVA and the chi-square analysis.
- ANOVA identifies whether at least one group mean differs but does not identify the specific pairs that differ. Post-hoc testing would be needed.
- Statistical significance does not automatically imply a large or commercially important effect.
- Outliers were flagged using the IQR method but were not automatically removed.
